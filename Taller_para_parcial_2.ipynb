Proyecto 2: En un estudio académico se desea analizar el comportamiento de las calificaciones de un grupo de estudiantes. Se asume que las notas siguen aproximadamente una distribución normal.

Pregunta de investigación: ¿Las calificaciones de los estudiantes pueden considerarse normalmente distribuidas y cuál es su comportamiento promedio?

Genere un conjunto de datos simulados de tamaño n=50 provenientes de una distribución normal con media 4.0 y desviación estándar 0.5.
Construya un DataFrame con estos datos.
Realice un análisis descriptivo: Media, Mediana, Desviación estándar, Mínimo y máximo
Visualice los datos (histograma).
Aplique la prueba de normalidad de Shapiro-Wilk.
Interprete el resultado con un nivel de significancia del 5%.
Responda la pregunta de investigación.


```python
import numpy as np
import pandas as pd

np.random.seed(123)   # para guardar la semilla

datos = np.random.normal(loc=4.0, scale=0.5, size=50) # simula 50 valores de la normal con media 4 y desviacion estandar 0.5

df = pd.DataFrame({      # para tranformarlo en un dataframe pues era un numpy
    "nota": datos
})
df
```

```python
df.describe()
```

```python
media = df["nota"].mean()
mediana = df["nota"].median()
desviacion_estandar = df["nota"].std()
minimo = df["nota"].min()
maximo = df["nota"].max()

print(f"Media: {media}")
print(f"Mediana: {mediana}")
print(f"Desviación estándar: {desviacion_estandar}")
print(f"Mínimo: {minimo}")
print(f"Máximo: {maximo}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import norm

# Datos
data = df["nota"].dropna()

# Parámetros de la normal
mu, std = data.mean(), data.std()

# Configuración estética (rosado bonito 🌸)
sns.set(style="whitegrid")
plt.figure(figsize=(10,6))

# Histograma (densidad)
sns.histplot(data, bins=20, stat="density",
             color="#f7a1c4", edgecolor="white", alpha=0.6)

# KDE (estimación por kernel)
sns.kdeplot(data, color="#d63384", linewidth=2, label="KDE")

# Curva normal teórica
x = np.linspace(data.min(), data.max(), 1000)
plt.plot(x, norm.pdf(x, mu, std),
         color="#ff69b4", linestyle="--", linewidth=2,
         label="Normal teórica")

# Detalles
plt.title("Distribución de 'nota' 🌸", fontsize=14)
plt.xlabel("Nota")
plt.ylabel("Densidad")
plt.legend()

plt.show()
```

```python
from scipy.stats import shapiro
shapiro(df["nota"])
```

como el valor p es mayor que 0.05 no se rechaza la hipotesis de normalida con una confianza del 95%

Proyecto 3: Una institución educativa desea comparar el rendimiento académico de dos grupos de estudiantes (Grupo A y Grupo B). Se sospecha que podría existir una diferencia en las calificaciones promedio entre ambos grupos.

Pregunta de investigación: ¿Existe evidencia estadística para afirmar que las medias de las calificaciones de los dos grupos son diferentes?

Genere dos muestras aleatorias de tamaño n=40 cada una:
Grupo A: distribución normal con media 4.0 y desviación estándar 0.4
Grupo B: distribución normal con media 4.3 y desviación estándar 0.5
Construya un DataFrame que contenga ambas muestras.
Realice un análisis descriptivo para cada grupo:
Media
Desviación estándar
Mínimo y máximo
Visualice los datos (histogramas o gráficos comparativos).
Importe el siguiente módulo estadístico:
from scipy import stats
Aplique:
•	Prueba de normalidad de Shapiro-Wilk para cada grupo
•	Prueba t de Student para comparar medias (stats.ttest_ind)
•	Use un nivel de significancia del 5%.
•	Interprete los resultados y responda la pregunta de investigación.


```python
from scipy import stats
```

```python
import numpy as np
import pandas as pd

np.random.seed(123)

grupo_A = np.random.normal(loc=4.0, scale=0.4, size=40)
grupo_B = np.random.normal(loc=4.3, scale=0.5, size=40)
```

```python
df = pd.DataFrame({
    "A": grupo_A,
    "B": grupo_B
})

t_stat, p_val = stats.ttest_ind(df["A"], df["B"])

print("Estadístico t:", t_stat)
print("Valor p:", p_val)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# DataFrame
df = pd.DataFrame({
    "A": grupo_A,
    "B": grupo_B
})

# Estilo rosado 🌸
sns.set_theme(style="whitegrid")

plt.figure(figsize=(10,6))

# Histograma A
sns.histplot(df["A"], bins=20, stat="density",
             color="#f7a1c4", alpha=0.5, label="Grupo A", kde=True)

# Histograma B
sns.histplot(df["B"], bins=20, stat="density",
             color="#ff69b4", alpha=0.5, label="Grupo B", kde=True)

# Detalles
plt.title("Histogramas de A y B 🌸", fontsize=14)
plt.xlabel("Valores")
plt.ylabel("Densidad")
plt.legend()

plt.show()
```

## Se rechaza que las calificaciones de ambos grupos sean iguales con una confianza del 95%
