Escritura de archivos desde python

```python
# Crear y escribi mun archivo de texto
with open("archivo.txt", "w") as f:
    f.write("Hola, este es un archivo de texto.\n")
    f.write("Segunda linea.\n")
```

Con el argumento "w" se crea el archivo si no existe. Tambien sobreescribe el archivo si ya existe, borrando todo su contenido previo

Archivos txt.

```python
with open("archivo.txt", "r") as f:
     print(f.read())
```

Con archivos .csv

```python
import csv

datos=[
    ["Nombre" , " Edad", "Ciudad"],
    ["Ana", 25, "Madrid"],
    ["Juan", 30, "Barcelona"],
    ["Camila", 27, "Valencia"]
]

with open ("datos.csv", "w", newline="") as f:
  writer = csv.writer(f)
  writer.writerows(datos)
```

```python
with open("datos.csv", "r") as f:
     print(f.read())
```
